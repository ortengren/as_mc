```python
import numpy as np
import ase
import ase.io
```

```python
bframes = ase.io.read("xyz_files/benzenes.xyz", ":")
ellframes = ase.io.read("xyz_files/ellipsoids_with_axes.xyz", ":")
duped_ell_frames = ase.io.read("xyz_files/duped_ellipsoids_with_axes.xyz", ":")
```

```python
energy_pa = [frame.info["energy_pa"] for frame in bframes]
mean_energy_pa = np.mean(energy_pa)
print(mean_energy_pa)
```

```python
for i, frame in enumerate(ellframes):
    frame.info["energy_pa"] = energy_pa[i]
    frame.info["intermolecular_energy_pa"] = energy_pa[i] - mean_energy_pa
    duped_ell_frames[i].info["energy_pa"] = energy_pa[i]
    duped_ell_frames[i].info["intermolecular_energy_pa"] = energy_pa[i] - mean_energy_pa
```

```python
ase.io.write("xyz_files/ellipsoids_with_axes.xyz", ellframes)
ase.io.write("xyz_files/duped_ellipsoids_with_axes.xyz", duped_ell_frames)
```
